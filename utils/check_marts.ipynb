```python
import io
import pandas as pd
from create_s3_client import BUCKET_NAME, s3_client

pd.set_option(
    'display.max_rows',
    None
    )
pd.set_option(
    'display.width',
    1000
    )


def read_gold_marts(mart_name: str) -> pd.DataFrame:
    key = f"gold/marts/{mart_name}.parquet"
    obj = s3_client.get_object(
        Bucket=BUCKET_NAME,
        Key=key
        )
    return pd.read_parquet(
        io.BytesIO(
            obj["Body"].read()
            )
        )
```

```python
print(
    read_gold_marts(
        "mart_sales_by_category"
        )
    )
df_cat = read_gold_marts("mart_sales_by_category")
print(df_cat)
```
